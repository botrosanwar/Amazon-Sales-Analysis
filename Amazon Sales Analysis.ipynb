## 1. Importing the used libraries

```python
import numpy as np
import pandas as pd
```

## 2. Read the CSV file

```python
df = pd.read_csv("amazon.csv")
```

## 3. Viewing and inspecting data (EDA)

```python
# Checking the shape
print(f"Dataset Shape: {df.shape}")

# Checking the data types and non-null counts
print("\nDataset info:")
df.info()

# Checking summary statistics
print("\nSummary Statistics:")
display(df.describe(include= "all"))

# Checking the count of missing values
print(f"\nThe count of missing values:\n{df.isnull().sum()}")

# Printing the first 5 records
print("\nThe first 5 records:")
display(df.head())
```

## 4. Data Cleaning & Wrangling

```python
# Remove unnecessary characters like "₹", "%", "," from numbers and cost them to float/int

df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)
```

```python
# Fixing rating and rating_count columns

df['rating'] = df['rating'].str.replace('|', '', regex=False)
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')  # it converts everything to float, and if it encounter something that cannot be converted, it replace it with "NaN" value

df['rating_count'] = df['rating_count'].str.replace(',', '')
df['rating_count'] = pd.to_numeric(df['rating_count'], errors= 'coerce')
```

```python
# handling missing values using numpy
avg_rating = np.nanmean(df['rating'])
df['rating'] = df['rating'].fillna(avg_rating)

avg_rating_count = np.nanmean(df['rating_count'])
df['rating_count'] = df['rating_count'].fillna(avg_rating_count)
```

```python
# Extract the main category (first item before the first '|')

df['category'] = df['category'].astype(str)
df['main_category'] = df['category'].apply(lambda x: x.split('|')[0])
```

```python
# Now let's repeat the inspection (EDA) to see what have changed

# Checking the shape
print(f"Dataset Shape: {df.shape}")      # The columns increased by 1 "Main categories", so becoming 17 instead of 16

# Checking the data types and non-null counts
print("\nDataset info:")
df.info()        # Not all datatype is object, some has been converted into float64, and there are no missing values

# Checking summary statistics
print("\nSummary Statistics:")
display(df.describe(include= "all"))      # There is a quantitative (numerical) stats, not only qualitative (categorical)

# Checking the count of missing values
print(f"\nThe count of missing values:\n{df.isnull().sum()}")

# Printing the first 5 records
print("\nThe first 5 records:")
display(df.head())
```

## 5. Feature Engineering

```python
df.columns
```

```python
# First, make a column called "Discount amount" which refer to the actual savings in Currency (Actual - Discounted)

df['discount_amount'] = df['actual_price'] - df['discounted_price']

# Second, make a column called "price_per_rating" which refer to price paid per rating point (Lower ratio = higher relative value for rating)

df['price_per_rating'] = np.round(df['discounted_price'] / df['rating'], 2)       # Round to two decimal numbers

# Third, make a column called "rating_tier" which categorizes products into qualitative performance buckets

conditions = [(df['rating'] >= 4.5),
             (df['rating'] >= 4.0) & (df['rating'] < 4.5),
             (df['rating'] >= 3.5) & (df['rating'] < 4.5)]

choices = ['Excellent', 'Good', 'Average']

df['rating_tier'] = np.select(conditions, choices, default='Low')       # constructs a new array by selecting elements from choices based on multiple specified conditions

print("Engineered features:")
print(df[['product_name', 'actual_price', 'discounted_price', 'discount_amount', 'rating', 'price_per_rating', 'rating_tier']].head())
print("\nDistribution of Rating Tiers:\n")
print(df['rating_tier'].value_counts())         # The count of every tier
```

## 6. Exploratory Data Analysis (EDA) & NumPy Business Insights

```python
df.columns
```

```python
# First core business question is about "Category Performance", Which main categories generate the highest average discounts and customer engagement?

category_summary = df.groupby('main_category').agg(
    product_count = ('product_id', 'count'),
    avg_discount_percentage = ('discount_percentage', 'mean'),
    avg_rating = ('rating', 'mean'),
    total_reviews = ('rating_count', 'sum')
).reset_index().sort_values(by = 'product_count', ascending = False).round(2)

print('The main categories performance:\n')
print(category_summary)
```

```python
# Second core business question is about "Discount vs. Rating Correlation", Do higher discounts lead to better customer ratings, or do heavily discounted items signal poor quality?

corr_matrix =  df[['discount_percentage', 'rating', 'rating_count', 'discount_amount']].corr()

print(f"Correlation Matrix:\n{corr_matrix}")

# Print specific statistical takeaway
print(f"\nThe specific correlation between (Discount % vs. Rating): {np.round(corr_matrix.iloc[0,1], 4)}")
```

```python
# Third core business question is about "Rating Tier Revenue Impact", How are products distributed across rating tiers (Exceptional, Good, Average, Low), and what is the median total volume of reviews for each tier?

tier_summary = df.groupby('rating_tier').agg(
    total_products=('product_id', 'count'),
    median_price=('discounted_price', 'median'),
    avg_rating=('rating', 'mean'),
    median_reviews=('rating_count', 'median')
).reindex(['Excellent', 'Good', 'Average', 'Low'])

print("Summary by Rating Tier:\n")
print(tier_summary)
```

```python
# Third core business question is about "Outlier Detection via NumPy", Identify extreme discount outliers using NumPy percentile bounds ($IQR$ method).

q75, q25 = np.percentile(df['discount_percentage'], [75, 25])
iqr = q75 - q25
upper_bound = q75 + (1.5 * iqr)
lower_bound = q25 - (1.5 * iqr)

outliers = df[df['discount_percentage'] > upper_bound]
print(f"\n--- Discount Outlier Detection ---")
print(f"Upper Bound threshold for Discount %: {upper_bound:.2f}%")
print(f"Number of Extreme Discount Outliers: {len(outliers)}")
```
